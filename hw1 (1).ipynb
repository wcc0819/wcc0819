## 问题 1

根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$
3. $T(1)=1;T(n)=T(2n/3)+1$ for $n>1$
4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$
5. $T(n)=1 \ for\ n \le 2 ; T(n)=T(\sqrt{n})+1 \ for \ n>2$

````python
问题 1：T(1) = 1；T(n) = T(n/4) + 1，n > 1

步骤 1	识别参数
a = 1（子问题个数），b = 4（规模缩小倍数），f(n) = 1（额外代价）

步骤 2	计算基准值
log_b(a) = log₄(1) = 0
因此 n^(log_b a) = n⁰ = 1

步骤 3	判断情况
f(n) = 1 = Θ(n⁰) = Θ(n^(log_b a))
=> f(n) 与基准值同阶，满足主定理 情况 2

步骤 4	套用公式
T(n) = Θ(n^(log_b a) · log n) = Θ(n⁰ · log n) = Θ(log n)

最终答案：T(n) = Θ(log n)

问题 2：T(1) = 1；T(n) = 3T(n/3) + n²，n > 1

步骤 1	识别参数
a = 3，b = 3，f(n) = n²

步骤 2	计算基准值
log_b(a) = log₃(3) = 1
因此 n^(log_b a) = n¹ = n

步骤 3	判断情况
f(n) = n²，基准值 = n
f(n) = Ω(n^(1+ε))，取 ε = 1，n² >> n¹，满足主定理 情况 3
验证正则条件：a·f(n/b) = 3·(n/3)² = n²/3 ≤ (1/3)·n²，取 c = 1/3 < 1  ✓

步骤 4	套用公式
T(n) = Θ(f(n)) = Θ(n²)

最终答案：T(n) = Θ(n²)

问题 3：T(1) = 1；T(n) = T(2n/3) + 1，n > 1

步骤 1	识别参数
a = 1，b = 3/2，f(n) = 1

步骤 2	计算基准值
log_b(a) = log_(3/2)(1) = 0
因此 n^(log_b a) = n⁰ = 1

步骤 3	判断情况
f(n) = 1 = Θ(n⁰) = Θ(n^(log_b a))
=> 满足主定理 情况 2

步骤 4	套用公式
T(n) = Θ(n^(log_b a) · log n) = Θ(log n)

最终答案：T(n) = Θ(log n)

问题 4：T(1) = 1；T(n) = 5T(n/4) + n，n > 1

步骤 1	识别参数
a = 5，b = 4，f(n) = n

步骤 2	计算基准值
log_b(a) = log₄(5) ≈ 1.16
因此 n^(log_b a) = n^(log₄5) ≈ n^1.16

步骤 3	判断情况
f(n) = n，基准值 = n^(log₄5) ≈ n^1.16
f(n) = O(n^(log₄5 - ε))，取 ε ≈ 0.16
f(n) 增长慢于基准值，满足主定理 情况 1

步骤 4	套用公式
T(n) = Θ(n^(log_b a)) = Θ(n^(log₄5))
注：log₄5 = ln5/ln4 ≈ 1.16，为无理数指数，保留精确形式

最终答案：T(n) = Θ(n^(log₄5))

问题 5：T(n) = 1（n ≤ 2）；T(n) = T(√n) + 1（n > 2）  ← 换元法
本题不能直接使用主定理，因为递归参数为 √n，不符合 n/b 形式，需先进行变量替换。

步骤 1	换元
令 m = log₂(n)，即 n = 2^m
则 √n = (2^m)^(1/2) = 2^(m/2)
原式变为：T(2^m) = T(2^(m/2)) + 1

步骤 2	令 S(m) = T(2^m)
则递归式化为：S(m) = S(m/2) + 1
这是标准的主定理形式！

步骤 3	对 S(m) 使用主定理
a = 1，b = 2，g(m) = 1
log_b(a) = log₂(1) = 0，m^(log_b a) = m⁰ = 1
g(m) = 1 = Θ(1) = Θ(m^(log_b a))，满足情况 2
=> S(m) = Θ(m^0 · log m) = Θ(log m)

步骤 4	回代 m = log₂(n)
T(n) = S(log₂n) = Θ(log(log₂n)) = Θ(log log n)

最终答案：T(n) = Θ(log log n)
````

## 问题 2

观察下方的分治算法（divide-and-conquer algorithm）的伪代码, 回答下面问题

```latex
DoSomething(A,p,r)
-----
n := r-p+1
if n=2 and A[p]>A[r] then
    swap A[p] and A[r]
else if n >= 3 then
    m = ceil(2n/3)
    DoSomething(A,p,p+m-1)
    DoSomething(A,r-m+1,r)
    DoSomething(A,p,p+m-1)  
    
---
first call: DoSomething(A,1,n)
```

note：$ceil(2n/3)=\left\lceil {2n/3} \right\rceil$；$:=$ 表示赋值, 等价于 $\to$；A是一个包含n的整数元素的数组, 

1. 写出该算法时间复杂度的递归公式, 并求解其对应的渐进表示
2. 描述一下该算法的功能, 并判断是否是最高效的解决方案
3. \* 使用python编程实现上述算法或其对应的更高效版本的算法

idea:

````python
第一问：递归式推导
每次调用规模 n，产生 3 个规模为 ⌈2n/3⌉ 的子问题，额外只做一次比较/交换（O(1)）：

T(n) = 3T(2n/3) + 1

主定理：a=3，b=3/2，f(n)=1，基准值 = n^(log_{3/2} 3) ≈ n^2.71，f(n)=O(1) 满足情况1：

T(n) = Θ(n^(log_{3/2} 3)) ≈ Θ(n^2.709)


第二问：功能描述与效率判断
这个算法就是著名的 Stooge Sort（臭皮匠排序），功能是对数组排序。绝对不是最高效方案，甚至比冒泡排序（Θ(n²)）还慢，因为 2.71 > 2。实际运行数据也证实了这点：n=1000 时，Stooge 比归并排序慢 5556 倍。

第三问：代码实现:
"""
问题 2 解答：DoSomething 算法分析与实现
======================================
算法识别：Stooge Sort（臭皮匠排序）
递归式：T(n) = 3T(2n/3) + 1
渐近复杂度：Θ(n^(log_{3/2} 3)) ≈ Θ(n^2.709)
"""

import math
import random
import time




def do_something(A, p, r):
    """
    原题伪代码的 Python 实现（1-indexed 转 0-indexed）
    A : 数组（list），直接原地修改
    p : 起始下标（0-indexed）
    r : 结束下标（0-indexed）
    """
    n = r - p + 1

    if n == 2:
        # 基础情况：两元素，若逆序则交换
        if A[p] > A[r]:
            A[p], A[r] = A[r], A[p]

    elif n >= 3:
        m = math.ceil(2 * n / 3)          # 前 2/3 的长度
        do_something(A, p, p + m - 1)     # 排序前 2/3
        do_something(A, r - m + 1, r)     # 排序后 2/3
        do_something(A, p, p + m - 1)     # 再次排序前 2/3（保证整体有序）


def stooge_sort(A):
    """对外接口：排序整个数组"""
    arr = A[:]
    do_something(arr, 0, len(arr) - 1)
    return arr



def merge_sort(A):
    """归并排序，时间复杂度 Θ(n log n)"""
    if len(A) <= 1:
        return A[:]
    mid = len(A) // 2
    left = merge_sort(A[:mid])
    right = merge_sort(A[mid:])
    return _merge(left, right)


def _merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result


def analyze_recurrence():
    print("=" * 55)
    print("  递归式分析：T(n) = 3T(2n/3) + 1")
    print("=" * 55)

    a = 3
    b = 3 / 2       # 规模缩小因子：n / (2n/3) = 3/2
    f_n = "O(1)"

    log_ba = math.log(a, b)
    print(f"\n  参数识别：")
    print(f"    a = {a}（每次产生 3 个子问题）")
    print(f"    b = 3/2（子问题规模 = 2n/3 = n / (3/2)）")
    print(f"    f(n) = 1（基础比较/交换操作）")
    print(f"\n  计算基准值：")
    print(f"    log_b(a) = log_{{3/2}}(3) = ln3/ln(3/2) ≈ {log_ba:.4f}")
    print(f"    n^(log_b a) ≈ n^{log_ba:.4f}")
    print(f"\n  判断情况（主定理）：")
    print(f"    f(n) = O(1) << n^{log_ba:.4f}，满足情况 1")
    print(f"    => T(n) = Θ(n^(log_{{3/2}} 3)) ≈ Θ(n^{log_ba:.4f})")
    print(f"\n  结论：")
    print(f"    Stooge Sort 时间复杂度 ≈ Θ(n^2.71)")
    print(f"    远差于归并排序的 Θ(n log n)")
    print()




def verify_correctness():
    print("=" * 55)
    print("  正确性验证")
    print("=" * 55)

    test_cases = [
        [3, 1, 4, 1, 5, 9, 2, 6],
        [5, 4, 3, 2, 1],
        [1],
        [2, 1],
        [9, 8, 7, 6, 5, 4, 3, 2, 1, 0],
    ]

    all_pass = True
    for arr in test_cases:
        result = stooge_sort(arr)
        expected = sorted(arr)
        ok = result == expected
        if not ok:
            all_pass = False
        print(f"  输入: {arr}")
        print(f"  输出: {result}  {'✓' if ok else '✗'}")
        print()

    print(f"  所有测试 {'全部通过 ✓' if all_pass else '存在失败 ✗'}")
    print()




def benchmark():
    print("=" * 55)
    print("  性能对比：Stooge Sort vs Merge Sort")
    print("=" * 55)
    print(f"  {'规模 n':<10} {'Stooge(ms)':<16} {'Merge(ms)':<14} {'倍数'}")
    print(f"  {'-'*50}")

    for n in [50, 100, 200, 500, 1000]:
        arr = [random.randint(0, 10000) for _ in range(n)]

        # Stooge Sort
        t0 = time.perf_counter()
        for _ in range(3):
            stooge_sort(arr)
        stooge_ms = (time.perf_counter() - t0) / 3 * 1000

        # Merge Sort
        t0 = time.perf_counter()
        for _ in range(20):
            merge_sort(arr)
        merge_ms = (time.perf_counter() - t0) / 20 * 1000

        ratio = stooge_ms / merge_ms if merge_ms > 0 else float('inf')
        print(f"  {n:<10} {stooge_ms:<16.3f} {merge_ms:<14.4f} {ratio:.1f}x")

    print()
    print("  理论预测：当 n 翻倍，Stooge 约慢 2^2.71 ≈ 6.6 倍")
    print("           当 n 翻倍，Merge  约慢 2^1.00 ≈ 2.0 倍（含 log 增长）")
    print()


# ══════════════════════════════════════════════════════
# 主程序
# ══════════════════════════════════════════════════════

if __name__ == "__main__":
    analyze_recurrence()
    verify_correctness()
    benchmark()

    print("=" * 55)
    print("  总结")
    print("=" * 55)
    print("""
  1. 递归式：T(n) = 3T(2n/3) + 1
     渐近复杂度：Θ(n^(log_{3/2} 3)) ≈ Θ(n^2.709)

  2. 算法功能：对数组进行排序（Stooge Sort）
     是否最优：否，远差于归并排序 Θ(n log n)
               甚至差于冒泡排序 Θ(n²)
               仅具理论研究价值

  3. 高效替代：归并排序（Merge Sort）
               T(n) = 2T(n/2) + n => Θ(n log n)
    """)
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:

````python
第一问：分治算法 O(n log n)
把数组从中点一分为二，最大子数组只有三种可能：

完全在左半
完全在右半
跨越中点（从 mid 向两侧各扫一遍，O(n)）

递归式：T(n) = 2T(n/2) + O(n) → 主定理情况2 → O(n log n)

第二问：Kadane 算法 O(n)
动态规划核心思想：dp[i] = max(A[i], dp[i-1] + A[i])
"""
问题 3：最大利润股票问题

本质：最大子数组和（Maximum Subarray Sum）
"""

import time
import random



def max_crossing_subarray(A, lo, mid, hi):
    """求必须跨越 mid 的最大子数组，O(n)"""
    # 从 mid 向左扫，找左侧最大前缀和
    left_sum = float('-inf')
    total = 0
    left_idx = mid
    for i in range(mid, lo - 1, -1):
        total += A[i]
        if total > left_sum:
            left_sum = total
            left_idx = i

    # 从 mid+1 向右扫，找右侧最大前缀和
    right_sum = float('-inf')
    total = 0
    right_idx = mid + 1
    for i in range(mid + 1, hi + 1):
        total += A[i]
        if total > right_sum:
            right_sum = total
            right_idx = i

    return left_sum + right_sum, left_idx, right_idx


def max_subarray_dc(A, lo, hi):
    """
    分治求最大子数组
    返回 (最大和, 买入下标, 卖出下标)，下标均从 0 开始
    """
  
    if lo == hi:
        return A[lo], lo, lo

    mid = (lo + hi) // 2

   
    left_val,  left_lo,    left_hi    = max_subarray_dc(A, lo, mid)
    right_val, right_lo,   right_hi   = max_subarray_dc(A, mid + 1, hi)
    cross_val, cross_lo,   cross_hi   = max_crossing_subarray(A, lo, mid, hi)


    if left_val >= right_val and left_val >= cross_val:
        return left_val, left_lo, left_hi
    elif right_val >= left_val and right_val >= cross_val:
        return right_val, right_lo, right_hi
    else:
        return cross_val, cross_lo, cross_hi


def max_profit_dc(profits):
    """对外接口（分治版），返回 (最大收益, 买入日, 卖出日)，日期从第1天开始"""
    n = len(profits)
    if n == 0:
        return 0, -1, -1
    if n == 1:
        return profits[0], 1, 1

    val, lo, hi = max_subarray_dc(profits, 0, n - 1)
    return val, lo + 1, hi + 1   # 转为 1-indexed




def max_profit_kadane(profits):
    """
    Kadane 算法，返回 (最大收益, 买入日, 卖出日)，日期从第1天开始
    """
    if not profits:
        return 0, -1, -1

    best_sum = float('-inf')
    best_lo = best_hi = 0

    cur_sum = 0
    cur_lo = 0                        # 当前子数组的起始下标

    for i, x in enumerate(profits):
        cur_sum += x

        if cur_sum > best_sum:        # 更新全局最优
            best_sum = cur_sum
            best_lo  = cur_lo
            best_hi  = i

        if cur_sum < 0:               # 当前子数组已亏损，重置
            cur_sum = 0
            cur_lo  = i + 1

    return best_sum, best_lo + 1, best_hi + 1   # 转为 1-indexed




def test(profits, label=""):
    val_dc,  buy_dc,  sell_dc  = max_profit_dc(profits)
    val_kd,  buy_kd,  sell_kd  = max_profit_kadane(profits)

    ok = (val_dc == val_kd)
    tag = "✓" if ok else "✗ 结果不一致！"
    print(f"  {'测试' if not label else label}")
    print(f"  profits = {profits}")
    print(f"  分治结果：买入第{buy_dc}天，卖出第{sell_dc}天，最大收益 = {val_dc}")
    print(f"  Kadane  ：买入第{buy_kd}天，卖出第{sell_kd}天，最大收益 = {val_kd}  {tag}")
    print()


def divider(title):
    print(f"\n{'='*55}")
    print(f"  {title}")
    print('='*55 + "\n")



divider("1. 题目给出的例子验证")
test([3, 2, 1, -7, 5, 2, -1, 3, -1], "题目例子（期望：第5天买入，第8天卖出，收益=9）")


divider("2. 更多测试用例")
test([1, 2, 3, 4, 5],          "全为正数")
test([-1, -2, -3, -4, -5],     "全为负数")
test([-2, 1, -3, 4, -1, 2, 1, -5, 4], "LeetCode 经典用例（期望=6）")
test([5],                       "单元素")
test([2, -1, 2, 3, -9, 10],    "混合数组")


divider("3. 性能对比：O(n log n) vs O(n)")
print(f"  {'规模 n':<12} {'分治(ms)':<16} {'Kadane(ms)':<14} {'加速比'}")
print(f"  {'-'*50}")

for n in [1000, 10000, 100000, 500000]:
    arr = [random.randint(-100, 100) for _ in range(n)]

    t0 = time.perf_counter()
    for _ in range(5):
        max_profit_dc(arr)
    dc_ms = (time.perf_counter() - t0) / 5 * 1000

    t0 = time.perf_counter()
    for _ in range(50):
        max_profit_kadane(arr)
    kd_ms = (time.perf_counter() - t0) / 50 * 1000

    ratio = dc_ms / kd_ms if kd_ms > 0 else float('inf')
    print(f"  {n:<12} {dc_ms:<16.3f} {kd_ms:<14.4f} {ratio:.1f}x")

# ── 总结 ──
divider("4. 算法总结")
print("""  第一问：分治算法
  ─────────────────────────────────────────────
  核心：将问题拆分为左半、右半、跨越中点三种情况
  递归式：T(n) = 2T(n/2) + O(n)
  时间复杂度：O(n log n)（主定理情况 2）
  空间复杂度：O(log n)（递归栈）

  第二问：Kadane 算法（最优解）
  ─────────────────────────────────────────────
  核心：动态规划，dp[i] = max(A[i], dp[i-1]+A[i])
  时间复杂度：O(n)，单次线性扫描
  空间复杂度：O(1)，只需常数个变量
  相比分治快约数十倍（实测数据如上）
""")
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:

````python
"""
问题 4：局部最小值查找
======================
给定数组 A[1..n]，满足 A[1]>A[2] 且 A[n-1]<A[n]
局部最小：A[x] < A[x-1] 且 A[x] < A[x+1]

第一问：O(log n) 二分查找，找到一个局部最小
第二问：O(n)    线性扫描，找出所有局部最小
"""

# ══════════════════════════════════════════════════════════════
# 第一问：二分查找一个局部最小  O(log n)
# ══════════════════════════════════════════════════════════════
#
# 算法设计思路：
#
# 关键性质：由于 A[1]>A[2]，左端天然"向内下坡"
#           由于 A[n-1]<A[n]，右端天然"向内下坡"
#           => 数组内部必然存在至少一个局部最小
#
# 二分策略（每次缩小一半搜索范围）：
#   取中点 mid = (lo + hi) // 2
#
#   情况 1：A[mid-1] < A[mid]（左侧比中点小，即中点在"上坡"）
#           => 左侧 [lo, mid-1] 的右端(mid-1)比 mid 小
#              而左端(lo)进来时是下坡，所以 [lo, mid-1] 内必有局部最小
#           => 缩小范围到左半：hi = mid - 1
#
#   情况 2：A[mid+1] < A[mid]（右侧比中点小，即中点在"下坡"）
#           => 右侧 [mid+1, hi] 的左端(mid+1)比 mid 小
#              而右端(hi)进来时是下坡，所以 [mid+1, hi] 内必有局部最小
#           => 缩小范围到右半：lo = mid + 1
#
#   情况 3：A[mid] < A[mid-1] 且 A[mid] < A[mid+1]
#           => mid 本身就是局部最小，直接返回
#
# 递归式：T(n) = T(n/2) + O(1)  =>  T(n) = O(log n)  [主定理情况1]

def find_local_min_binary(A):
    """
    二分查找一个局部最小值
    A: 1-indexed 逻辑，传入 0-indexed 的 Python list
    返回 (值, 1-indexed位置)
    """
    n = len(A)
    lo, hi = 1, n - 2          # 只在内部元素中搜索（边界由题目条件保证不是最小）

    while lo <= hi:
        mid = (lo + hi) // 2

        left_bigger  = (A[mid - 1] > A[mid])   # 左邻 > 中点
        right_bigger = (A[mid + 1] > A[mid])   # 右邻 > 中点

        if left_bigger and right_bigger:
            # 两侧都比 mid 大，mid 就是局部最小
            return A[mid], mid + 1              # +1 转为 1-indexed

        elif not left_bigger:
            # A[mid-1] <= A[mid]，左侧更小，往左找
            hi = mid - 1
        else:
            # A[mid+1] <= A[mid]，右侧更小，往右找
            lo = mid + 1

    # 理论上由题目条件保证必能找到，不会到这里
    return None, -1




def find_all_local_min(A):
    """
    找出所有局部最小值
    返回 [(值, 1-indexed位置), ...]
    """
    result = []
    n = len(A)
    for i in range(1, n - 1):           # 只检查内部元素
        if A[i] < A[i - 1] and A[i] < A[i + 1]:
            result.append((A[i], i + 1))   # +1 转为 1-indexed
    return result


# ══════════════════════════════════════════════════════════════
# 验证与测试
# ══════════════════════════════════════════════════════════════

def divider(title):
    print(f"\n{'='*55}")
    print(f"  {title}")
    print('='*55 + "\n")

def test(A, label=""):
    one_val, one_pos = find_local_min_binary(A)
    all_mins = find_all_local_min(A)

    print(f"  {label}")
    print(f"  A = {A}")
    print(f"  二分查找（一个）：第{one_pos}个元素，值 = {one_val}")
    print(f"  线性扫描（全部）：{all_mins}")

    # 验证二分结果是否真的是局部最小
    idx = one_pos - 1
    ok = (A[idx] < A[idx - 1] and A[idx] < A[idx + 1])
    print(f"  验证：{'✓ 正确' if ok else '✗ 错误！'}")
    print()


divider("1. 题目例子验证")
test([9, 3, 7, 2, 1, 4, 5], "题目例子（局部最小应为 3 和 1）")

divider("2. 更多测试用例")
test([10, 1, 3, 2, 5, 6, 7],  "靠近左端的局部最小")
test([10, 8, 6, 4, 2, 3, 9],  "单个谷底")
test([9, 5, 6, 1, 7, 3, 4],   "多个局部最小：5, 1, 3")
test([8, 1, 9, 2, 7, 3, 5],   "多个局部最小：1, 2, 3")


divider("3. 二分过程演示（逐步追踪）")

def find_local_min_verbose(A):
    n = len(A)
    lo, hi = 1, n - 2
    step = 0
    print(f"  数组：{A}  (n={n})")
    print(f"  初始搜索范围：[{lo+1}, {hi+1}]（1-indexed）\n")

    while lo <= hi:
        step += 1
        mid = (lo + hi) // 2
        print(f"  第{step}步：lo={lo+1}, hi={hi+1}, mid={mid+1}, A[mid]={A[mid]}")
        print(f"         左邻 A[{mid}]={A[mid-1]},  右邻 A[{mid+2}]={A[mid+1]}")

        if A[mid-1] > A[mid] and A[mid+1] > A[mid]:
            print(f"         => 两侧均大，找到局部最小！")
            print(f"\n  结果：第{mid+1}个元素，值 = {A[mid]}，共 {step} 步")
            return
        elif A[mid-1] <= A[mid]:
            print(f"         => 左侧更小，搜索左半 [{lo+1}, {mid}]")
            hi = mid - 1
        else:
            print(f"         => 右侧更小，搜索右半 [{mid+2}, {hi+1}]")
            lo = mid + 1
        print()

find_local_min_verbose([9, 3, 7, 2, 1, 4, 5])


divider("4. 性能对比")
import time, random

def make_valid_array(n):
    """生成满足题目条件的数组：A[0]>A[1] 且 A[n-2]<A[n-1]"""
    while True:
        A = random.sample(range(1, n * 10), n)
        if A[0] > A[1] and A[-2] < A[-1]:
            return A

print(f"  {'规模 n':<12} {'二分(μs)':<16} {'线性(μs)':<14} {'加速比'}")
print(f"  {'-'*50}")

for n in [1000, 10000, 100000, 1000000]:
    A = make_valid_array(n)

    t0 = time.perf_counter()
    for _ in range(1000):
        find_local_min_binary(A)
    bin_us = (time.perf_counter() - t0) / 1000 * 1e6

    t0 = time.perf_counter()
    for _ in range(100):
        find_all_local_min(A)
    lin_us = (time.perf_counter() - t0) / 100 * 1e6

    ratio = lin_us / bin_us
    print(f"  {n:<12} {bin_us:<16.2f} {lin_us:<14.2f} {ratio:.1f}x")

print(f"\n  说明：线性扫描虽然 O(n) 但目标不同（找全部），两者不可直接比较优劣")

divider("5. 总结")
print("""  第一问：二分查找 O(log n)
  ──────────────────────────────────────────────
  核心：利用"必然存在局部最小"的单调性保证
        检查中点与左右邻居大小关系，每次淘汰一半
  递归式：T(n) = T(n/2) + O(1)  =>  O(log n)

  第二问：线性扫描找全部  O(n)
  ──────────────────────────────────────────────
  核心：遍历所有内部元素，逐一检查是否满足局部最小条件
  无法比 O(n) 更快（最坏情况所有元素都是局部最小）
""")
````

3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:

````python
"""复杂度分析

处理每个元素：O(log k)
处理 n 个元素：O(n log k)
最后对堆里 k 个元素排序输出：O(k log k)
总体：O(n log k)
空间复杂度：O(k) ← 这是关键，和 n 无关"""

import heapq

def method3_streaming_top_k(data_stream, k):
    """方法：数据量超过内存，维护大小为k的最大堆
    
    data_stream: 可迭代对象，模拟无法全部读入内存的数据流
    k: 要找的前k小的数
    """
    max_heap = []  # 用负数模拟最大堆
    
    for x in data_stream:
        if len(max_heap) < k:
            # 堆还没满，直接放
            heapq.heappush(max_heap, -x)
        else:
            # 堆满了，比较新元素和当前堆里最大的（即 -max_heap[0]）
            current_max = -max_heap[0]
            if x < current_max:
                # 新元素更小，有资格进来
                heapq.heappushpop(max_heap, -x)
                # heappushpop 比先 pop 再 push 更高效
            # 否则丢弃 x，什么都不做
    
    # 把堆里的负数转回正数，并排序输出
    result = sorted([-v for v in max_heap])
    return result


# 测试：用生成器模拟"超大数据流"
def big_data_generator(arr):
    """模拟一个数据流，每次只能读一个数，无法整体载入"""
    for x in arr:
        yield x


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
print("方法结果:", method3_streaming_top_k(big_data_generator(arr), 3))
# 输出: [1, 2, 3]
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:

````python
"""总时间复杂度是 O(n)，空间复杂度是 O(n)（最坏情况下所有星球都入栈，比如全是正数或全是负数）。"""

def planet_collision(A):
    """
    模拟星球碰撞过程，返回最终存活的星球列表。
    
    思路：用栈维护已经"安定下来"的星球。
    遍历每个新星球时，判断它是否会与栈顶发生碰撞，
    通过 while 循环处理连锁碰撞。
    """
    stack = []  # 栈，存放已经安定下来的星球
    
    for x in A:
        # alive 标记当前星球 x 是否在所有碰撞后还活着
        alive = True
        
        # 只有 "x 向左 且 栈顶向右" 才可能发生碰撞
        # 用 while 是因为 x 可能连续撞碎多个栈顶星球
        while alive and x < 0 and stack and stack[-1] > 0:
            top = stack[-1]  # 栈顶星球（正数，向右走）
            
            if top < -x:
                # 栈顶质量小于 x 的质量，栈顶被撞碎
                # x 还活着，继续看新的栈顶
                stack.pop()
                # 注意：alive 保持 True，循环会继续判断
            elif top == -x:
                # 质量相等：根据题目的不对称规则
                # 向左走的 x 消失，向右走的 top 保留
                # 所以栈顶不动，x 死掉
                alive = False
            else:  # top > -x
                # 栈顶质量更大，x 被撞碎
                alive = False
        
        # 循环结束后，如果 x 还活着，就把它加入栈
        if alive:
            stack.append(x)
    
    return stack


# 测试题目给的两个例子
test1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
print(f"测试1输入: {test1}")
print(f"测试1输出: {planet_collision(test1)}")
# 期望: [-3, -6, 2, 8, 9, 1]

test2 = [23, -8, 9, -3, -7, 9, -23, 22]
print(f"测试2输入: {test2}")
print(f"测试2输出: {planet_collision(test2)}")
# 期望: [23, 22]
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

````python
"""DP 的核心思想是避免重复计算。我们注意到，在做"第 i 块石头选不选"的决策时，真正影响后续决策的状态其实只有两个量：当前考虑到第几块石头，以及当前剩余的承重容量。只要这两个量相同，无论之前是怎么走过来的，剩下的最优决策都一样。这就是 DP 能成立的"无后效性"。"""



def knapsack_2d(items, capacity):
    """
    0-1 背包问题的二维 DP 解法。
    
    items: 列表，每个元素是 (重量, 价值) 元组
    capacity: 背包最大承重
    返回: 最大价值
    """
    n = len(items)
    # dp[i][w] 表示前 i 件物品在容量 w 下的最大价值
    # 多开一行一列方便处理边界（i=0 或 w=0 的情况）
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    
    for i in range(1, n + 1):
        weight, value = items[i - 1]  # 注意 items 下标从 0 开始
        for w in range(capacity + 1):
            # 默认：不拿第 i 件
            dp[i][w] = dp[i - 1][w]
            # 如果能装下，比较"拿"和"不拿"
            if w >= weight:
                dp[i][w] = max(dp[i][w], dp[i - 1][w - weight] + value)
    
    return dp[n][capacity]


# 测试
items = [(5, 10), (4, 40), (6, 30), (3, 50)]
capacity = 9
print(f"最大价值: {knapsack_2d(items, capacity)}")
# 输出: 90

"""复杂度分析 
无论是二维还是一维写法，时间复杂度都是 O(nW)，其中 n 是物品数量，W 是背包容量。这是因为我们要填满整张 DP 表（或一维数组的 n 次更新），每个格子的计算是 O(1)。
空间复杂度上，二维写法是 O(nW)，一维滚动数组优化后是 O(W)。"""
````

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:

````python
"""Morris 中序遍历的逻辑：对当前节点 cur，如果它没有左子树，那么直接访问它本身（这一步对应"左为空时根就是当前要访问的节点"），然后转向右子树；如果它有左子树，就找到左子树中最右边的节点（也就是中序遍历下 cur 的前驱节点），把这个前驱节点的右指针临时指向 cur，然后进入左子树继续处理。当我们沿着左子树一路向下，再通过这些临时建立的"右指针线索"绕回来时，就能恢复到 cur 进行访问，并把临时指针拆掉，恢复树的原始结构."""





def recover_tree_v3(root):
    """
    解法：Morris 中序遍历，O(1) 空间复杂度。
    
    核心思想：利用叶子节点空闲的右指针建立"线索"，
    遍历完成后再拆除，整体不使用递归或栈。
    """
    first = second = prev = None
    cur = root
    
    while cur:
        if cur.left is None:
            # 情况一：没有左子树
            # 此时 cur 就是中序序列里下一个要访问的节点
            if prev and prev.val > cur.val:
                if first is None:
                    first = prev
                second = cur
            prev = cur
            cur = cur.right  # 转向右子树
        else:
            # 情况二：有左子树
            # 找到左子树中最右节点，即 cur 在中序序列中的前驱
            predecessor = cur.left
            while predecessor.right and predecessor.right != cur:
                predecessor = predecessor.right
            
            if predecessor.right is None:
                # 第一次到达这个前驱：建立线索，让它指回 cur
                # 这样我们处理完左子树后能"绕回来"
                predecessor.right = cur
                cur = cur.left  # 进入左子树
            else:
                # 第二次到达这个前驱：说明左子树已经处理完了
                # 此时拆除线索，并访问 cur
                predecessor.right = None
                if prev and prev.val > cur.val:
                    if first is None:
                        first = prev
                    second = cur
                prev = cur
                cur = cur.right  # 转向右子树
    
    # 交换两个错位节点的值
    first.val, second.val = second.val, first.val

"""Morris 遍历的时间复杂度仍然是 O(n)，虽然看起来每个有左子树的节点要找两次前驱，但通过摊还分析可以证明每条边最多被走两次，总时间还是线性的。空间复杂度真正做到了 O(1)，没有递归栈也没有显式栈，只用了几个指针变量。"""
````

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:

````python
from collections import deque

def sum_numbers_bfs(root):
    """
    用 BFS（迭代）方式实现。
    用队列同时维护节点和该节点对应的累积数字。
    """
    if root is None:
        return 0
    
    total = 0
    # 队列中每个元素是一个二元组：(节点, 该节点对应的累积数字)
    queue = deque([(root, root.val)])
    
    while queue:
        node, current = queue.popleft()
        
        # 到达叶节点：累加这条路径的数字
        if node.left is None and node.right is None:
            total += current
            continue  # 叶节点不再向下扩展
        
        # 非叶节点：把孩子和"扩展后的数字"加入队列
        if node.left:
            queue.append((node.left, current * 10 + node.left.val))
        if node.right:
            queue.append((node.right, current * 10 + node.right.val))
    
    return total
"""复杂度：时间复杂度都是 O(n)，空间复杂度O(n)。"""
````
